**Q1**. Load the given retail sales dataset into a Pandas DataFrame and display the first 5 rows along with the
dataset information to understand its structure.

**Ans -**


```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv("retail_dataset .csv")

# Display the first 5 rows
print("First 5 rows of the dataset:")
display(df.head())

# Display dataset information
print("Dataset Information:")
df.info()
```

**Q2.** Inspect the dataset and identify missing values in each column. Display the total number of missing
values present in the dataset.

**Ans -**


```python
# Check missing values in each column
print("Missing values in each column:")
print(df.isnull().sum())

# Display total missing values
total_missing = df.isnull().sum().sum()

print("\nTotal missing values in the dataset:", total_missing)
```

**Q3.** Replace missing values in the Quantity and Price columns with their respective mean values using
Pandas.

**Ans -**


```python
# Calculate the mean of Quantity and Price
quantity_mean = df["Quantity"].mean()
price_mean = df["Price"].mean()

# Replace missing values with the respective means
df["Quantity"] = df["Quantity"].fillna(quantity_mean)
df["Price"] = df["Price"].fillna(price_mean)

# Display the updated columns
print("Updated Quantity and Price columns:")
display(df[["Quantity", "Price"]].head(10))
```

**Q4.** Remove all rows where the Product Category or Region column contains missing values.

**Ans -**


```python
# Remove rows with missing Product Category or Region
df = df.dropna(subset=["Product Category", "Region"])

# Reset the index
df = df.reset_index(drop=True)

# Display the updated dataset
print("Dataset after removing missing categories and regions:")
display(df)

print("Number of rows after cleaning:", len(df))
```

**Q5.** Create a new column called Revenue using NumPy, where:

**Revenue = Quantity × Price**

Display the updated dataset with the new column.

```python
# Calculate Revenue using NumPy
df["Revenue"] = np.multiply(df["Quantity"], df["Price"])

# Display the updated dataset
display(df.head())
```

**Q6.** Using NumPy, calculate the total revenue generated from all transactions in the dataset.

**Ans -**



```python
# Calculate total revenue using NumPy
total_revenue = np.sum(df["Revenue"].to_numpy())

print("Total Revenue:", total_revenue)
```

**Q7.** Group the dataset by Product Category and calculate the total revenue generated by each category.

**Ans -**


```python
# Group by Product Category and calculate total revenue
category_revenue = df.groupby("Product Category")["Revenue"].sum()

# Display the result
print("Total revenue by Product Category:")
print(category_revenue)
```

**Q8.** Based on the revenue calculated for each product category, identify the top 3 categories with the
highest revenue and the bottom 3 categories with the lowest revenue.

**Ans -**


```python
# Calculate total revenue by category
category_revenue = df.groupby("Product Category")["Revenue"].sum()

# Sort revenue from highest to lowest
sorted_revenue = category_revenue.sort_values(ascending=False)

# Top 3 categories
top_3 = sorted_revenue.head(3)

# Bottom 3 categories
bottom_3 = sorted_revenue.tail(3).sort_values()

print("Top 3 categories with the highest revenue:")
print(top_3)

print("\nBottom 3 categories with the lowest revenue:")
print(bottom_3)
```

**Q9.** Group the dataset by Region and calculate the total revenue generated in each region. Identify the
region with the highest revenue and the region with the lowest revenue.

**Ans -**


```python
# Group by Region and calculate total revenue
region_revenue = df.groupby("Region")["Revenue"].sum()

# Display revenue by region
print("Total revenue by Region:")
print(region_revenue)

# Find the highest and lowest revenue regions
highest_region = region_revenue.idxmax()
lowest_region = region_revenue.idxmin()

print("\nRegion with the highest revenue:", highest_region)
print("Region with the lowest revenue:", lowest_region)
```

**Q10.** Extract the month from the Date column using Pandas and group the dataset by month to calculate the
total monthly revenue. Also, calculate the mean, median, and standard deviation of the Revenue column
using NumPy.

**Ans -**


```python
# Convert the Date column into datetime format
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True)

# Extract the month
df["Month"] = df["Date"].dt.month

# Calculate total monthly revenue
monthly_revenue = df.groupby("Month")["Revenue"].sum()

print("Total monthly revenue:")
print(monthly_revenue)

# Convert Revenue into a NumPy array
revenue_array = df["Revenue"].to_numpy()

# Calculate mean, median, and standard deviation using NumPy
mean_revenue = np.mean(revenue_array)
median_revenue = np.median(revenue_array)
std_revenue = np.std(revenue_array)

print("\nMean Revenue:", mean_revenue)
print("Median Revenue:", median_revenue)
print("Standard Deviation of Revenue:", std_revenue)
```
